## ¿Qué es una matriz mágica? 🔮

Es una matriz cuadrada N×N donde la suma de cada fila, cada columna y las dos diagonales es siempre el mismo número (llamado constante mágica).
Por ejemplo, para N=3:

```
    2  7  6
    9  5  1
    4  3  8
```
 Todas las filas, columnas y diagonales suman 15.
 
### ¿Qué funciones necesito? (Pseudocódigo) 💻
---
* Una función que me diga cual es el valor de la suma mágica |x|
* Una función que sume las filas |x|
* Una función que sume las columnas |x|
* Una función que sume la diagonal principal |x|
* Una función que sume la diagonal secundaria |x|
* Una función que compare todas las funciones anteriores |x|
* Una función que coordine todo lo anterior y verifique si se cumple que la matriz es mágica (True) o no (False) |x|


````python
# Funciones que necesito

# --- Matriz de prueba ---
matriz = [
    [2, 7, 6],
    [9, 5, 1],
    [4, 3, 8]
]

# Pista matriz mágica: fórmula → N * (N²+ 1) / 2

def calcular_suma_magica(matriz):
    N = len(matriz)
    return (N * (N**2 + 1) / 2)

print(f"El resultado de la suma mágica de la matriz de prueba es {calcular_suma_magica(matriz)}")
````

````python
def sumar_fila(matriz,fila):
    
    return sum(matriz[fila]) # dame el elemento en la posición que valga fila

print(f"El resultado de sumar la primera fila de la matriz es {sumar_fila(matriz, 0)}") 
print(f"El resultado de sumar la primera fila de la matriz es {sumar_fila(matriz, 1)}") 
print(f"El resultado de sumar la primera fila de la matriz es {sumar_fila(matriz, 2)}") 
````

````python
def sumar_columna(matriz,columna):
    total=0

    for i in range(len(matriz)):
        total += matriz[i][columna] # += sería igual que poner total = total + matriz[i][columna]
    return total

print(f"El resultado de sumar la primera columna de la matriz es {sumar_columna(matriz, 0)}")
print(f"El resultado de sumar la segunda columna de la matriz es {sumar_columna(matriz, 1)}")
print(f"El resultado de sumar la tercera columna de la matriz es {sumar_columna(matriz, 2)}") 
````

````python
def sumar_diagonal_principal(matriz):
    total=0
    for i in range(len(matriz)):
        total += matriz[i][i]
    return total

print(f"El resultado de sumar la diagonal principal de la matriz es {sumar_diagonal_principal(matriz)}")
````

````python
def sumar_diagonal_secundaria(matriz):
    total=0
    N = len(matriz)
    for i in range (N):
        total += matriz[i][N -1 - i] # matriz[i] accede a la fila i de la matriz
    return total

print(f"El resultado de sumar la diagonal secundaria de la matriz es {sumar_diagonal_secundaria(matriz)}")
````

Debemos que tener en cuenta también que una matriz mágica es cuadrada, es decir, tiene el mismo número de filas y de columnas:


````python
len(matriz) # número de filas
len(matriz[0]) # número de columnas

def es_cuadrada(matriz):
    if len(matriz) != len(matriz[0]):
        print(f" ⚠ Error de sistema la matriz de estudio no es cuadrada, por lo tanto tampoco puede ser mágica")
        return (False)
    return (True)   

print(f"La matriz de prueba es cuadrada: {es_cuadrada(matriz)}")
````

Perfecto, ya tenemos todas las piezas listas:

✅ sumar_columna

✅ sumar_diagonal_principal 

✅ sumar_diagonal_secundaria

✅ es_cuadrada

Ahora viene es_magica — la función principal que usa todas las anteriores para decidir si la matriz es mágica o no.
La lógica en pseudocódigo sería:
1. Calcular cuál es la suma mágica
2. Comprobar si es cuadrada
3. Comprobar que cada fila suma lo mismo
4. Comprobar que cada columna suma lo mismo
5. Comprobar que las dos diagonales suman lo mismo
6. Si todo cumple → devolver True, si no → devolver False

````python
''' Ahora necesito una función que compare todas los resultados de las funciones anteriores'''

def es_magica(matriz):
    # Calcular cual es la suma_magica

    suma_magica = calcular_suma_magica(matriz)

    # Comprobar si es cuadrada

    if not es_cuadrada(matriz):
        return(False)

    # Comprobar las filas y columnas (hay que hacer un bucle para recorrer toda la matriz)

    for i in range(len(matriz)):
        if sumar_fila(matriz, i) != suma_magica:
            return (False)
        
        elif sumar_columna(matriz, i) != suma_magica:
            return (False)
        
    # Comprobar diagonales (no hace falta bucle)

    if sumar_diagonal_principal(matriz) != suma_magica:
        return (False)
    if sumar_diagonal_secundaria(matriz) != suma_magica:
        return (False)
    
    return (True)    
````

````python
''' Matrices de prueba para comprobar que el código funciona'''

matriz_1 = [
    [2,7,6],
    [9,5,1],
    [4,3,8]
    ]

matriz_2 = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
    ]

''' Hacemos la prueba'''

print(f"La matriz_1 es mágica: {es_magica(matriz_1)}")     
print(f"La matriz_2 es mágica: {es_magica(matriz_2)}")
````
